# Tool functions

Claude only knows its training data. **Tool use** lets it ask *your code* for anything else:
Claude decides which tool to call and with what arguments, your code runs it and sends the result back.

The module project: Claude sets reminders ("remind me a week from Thursday"). Claude is unreliable
at knowing the exact time and at date arithmetic, and has no way to set a reminder, so it gets three tools:

| Tool | Why |
|---|---|
| `get_current_datetime` | the exact current time (this lesson) |
| `add_duration_to_datetime` | reliable date arithmetic |
| `set_reminder` | actually creates the reminder |

A **tool function** is a plain Python function that runs when Claude asks for it.

- Descriptive names for the function and its parameters
- Validate the input and raise an error when it is missing or invalid
- Write meaningful error messages: Claude sees them and can retry with fixed arguments

## The function

```python
from datetime import datetime


def get_current_datetime(date_format="%Y-%m-%d %H:%M:%S"):
    if not date_format:
        raise ValueError("date_format cannot be empty")
    return datetime.now().strftime(date_format)
```

## Call it directly

No API call yet: a tool function is ordinary code and can be tested on its own.

```python
print(get_current_datetime())
print(get_current_datetime("%H:%M"))
print(get_current_datetime("%A, %B %d, %Y"))
```

## Validation

An empty format is unlikely, but it shows the pattern. Later the error text is sent back to Claude
as the tool result, so it has to say what went wrong.

```python
try:
    get_current_datetime("")
except ValueError as error:
    print(f"Error: {error}")
```

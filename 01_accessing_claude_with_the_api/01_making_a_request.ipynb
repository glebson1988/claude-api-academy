# Making a request

Send one message to Claude with the Messages API and read the reply.

- `client.messages.create(model, max_tokens, messages)` is the core call
- `messages` is a list of `{"role": ..., "content": ...}` dicts
- the reply is a list of content blocks; the text is in the `text` blocks

## Setup

The client reads `ANTHROPIC_API_KEY` (and `ANTHROPIC_BASE_URL`, if set) from `.env` in the repo root.

```python
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()

client = Anthropic()
model = "claude-haiku-4-5"
```

## Send a request

```python
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence",
        }
    ],
)
```

## Read the text

The course uses `message.content[0].text`. The first block is text when no tools are used,
but type checkers flag it, so here all text blocks are joined instead.

```python
print("".join(block.text for block in message.content if block.type == "text"))
```

## What else comes back

`stop_reason` says why generation stopped (`end_turn`, `max_tokens`, `stop_sequence`, `tool_use`),
`usage` shows the tokens you pay for.

```python
print("model:", message.model)
print("stop_reason:", message.stop_reason)
print("input tokens:", message.usage.input_tokens, "| output tokens:", message.usage.output_tokens)
```
